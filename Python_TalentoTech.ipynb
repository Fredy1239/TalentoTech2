```python
#Calcular el promedio de edad de n personas
n = int(input("Ingrese el numero de personas:"))
k = 1 #Variable de control
suma = 0.0 #Almacenar los valores ingresados
while k <= n:  #Condicional para la ejecucion del ciclo
  edad = int(input("Ingrese la edad:")) #Se ingresa la edad
  suma = suma + edad #Se acumulan las edades
  k = k + 1 #Se actualiza el contador
print("El promedio de edad es:",suma/n)
```

```python
#Realizar un algoritmo para comparar las áreas de un triángulo y un cuadrado
#Ingresar los valores necesarios para calcular el área de cada objeto
base = int(input("Ingrese la base del triángulo:"))
altura = int(input("Ingrese la altura del triángulo:"))
lado = int(input("Ingrese el lado del cuadrado:"))
#Calcular las áreas de los objetos
areatriangulo=(base*altura)/2
areacuadrado=lado*lado
#Imprimir los resultados
print("El área del triángulo es:",areatriangulo)
print("El área del cuadrado es:",areacuadrado)
#Condicional para comparar los resultados
if areatriangulo >areacuadrado:
  print("El área del triángulo es mayor")
else:
  print("El área del cuadrado es mayor")
```

```python
#Algoritmo que grafique un triangulo rectangulo basado en un numero pedido al usuario
x =int(input("Ingrese numero entero:"))
k2 = 1 #Variable controlador
c2 = "*" #Gráfico
while k2 <= x: #Condicional que permite que la variable controlador siempre sea menor que la altura del triángulo
  print(c2*k2) #Se grafica en base a la variable controlador
  k2 =k2+1 #Se agranda el k2 buscando cumplir con la condicion inicial
```

```python
#Ejemplo ciclo anidado
i=1 #Variable Controla altura
n=int(input("Ingrese tamaño del triángulo"))
while i<=n:
  j=0 #Variable controla ancho
  while j<=n-i:
    print("*",end="")
    j=j+1
  print("")
  i=i+1
```

```python
#Importamos la libreria
import miceforest as mf
from sklearn.datasets import load_iris #Extraigo el dataset
from sklearn import datasets
import numpy as np
```

```python
iris = datasets.load_iris()
print(iris)
```

```python
df_iris = pd.DataFrame(data=iris.data,columns=iris.feature_names)
df_iris["target"] = iris.target
```

```python
df_iris.rename({"target":"species"}, inplace=True,axis=1)
df_iris
```

```python
df_iris["species"]=df_iris["species"].astype("category")
df_iris.info()
```

```python
#Asigna valores nulos aleatoriamente
#Quita valores válidos por nulos
iris_amp = mf.ampute_data(df_iris,perc=0.25,random_state=1991)
iris_amp
```

```python
#Crear Kernel
kds = mf.ImputationKernel(
    iris_amp, ## Nombre de DataFrame para entrenar el modelo
    random_state=1991 ## Valores aleatorios para entrenar el modelo
)

#Correr el algoritmo MICE durante 2 iteraciones
kds.mice(2)

iris_completo = kds.complete_data()
iris_completo
```

```python
# Boxplot

# La media es la linea en mitad de la caja y coincide con el Q50%
# El Q75% es la línea superior de la caja
# el Q25% es la línea inferior de la caja
# El tamaño de la caja representa la distribución de los datos
# El bigote es un intercuartil, valores por encima o por debajo de ellos son identificados como "outliers"
plt.figure(figsize=(10,5))
sns.set_theme(style="darkgrid", font_scale=1, palette= "bright")
sns.boxplot(
    x = hurto_a_persona["sexo"], # Tomo un campo categorico
    y = hurto_a_persona["edad"], # Tomó una variable numérica
    #palette="pastel")  # Configuro estilo de gráfica
)
plt.title(" Boxplot (Distribución) Edad de personas hurtadas por sexo") # Título de la gráfica
plt.xlabel("Sexo") # Titulo eje x
plt.ylabel("Edad") # Titulo eje y
plt.savefig("BOXPLOT.svg") # Formato SVG alta definición
plt.show()
```

```python
# Boxplot(Distribución) por estado civil y edad del hurto de personas
import matplotlib.pyplot as plt
import seaborn as sns
# Establecemos la representación visual de la gráfica
sns.set_theme(style="darkgrid", font_scale=1, palette= ["#800080"],font="sans-serif")

plt.figure(figsize=(10,5)) # Establecemos el tamaño de la gráfica
sns.boxplot( # Establecemos el tipo de gráfica
    x=hurto_a_persona["estado_civil"], # Variable categórica
    y =hurto_a_persona["edad"], # Variable Numérica
)
plt.title(" Boxplot (Distribución) Edad de personas hurtadas por estado civil") # Título de la gráfica
plt.xlabel("Estado Civil") # Titulo eje x
plt.ylabel("Edad") # Titulo eje y
plt.savefig("BOXPLOT2.svg") # Formato SVG alta definición
plt.show() # Imprimimos la gráfica
```

```python
# Gráfico sin apilar las categorías
sns.catplot( # Iniciamos el gráfico
    data = hurto_a_persona, # Se define el dataframe de origen de la información, se puede limitar el numero de variables a gráficar hurto_a_persona.head()    x = "estado_civil", # Valores del eje x
    x = "estado_civil", # Valores del eje x
    y = "edad", # Valores del eje y
    hue = "sexo", # Agrupa los resultados dependiendo de las variables categoricas
    kind = "bar" # Establecemos el tipo de gráfico
)
plt.title("Hurtos a personas por estado civil, sexo y edad ")
plt.xlabel("Estado Civil")
plt.ylabel("Edad")
plt.xticks(rotation = 90) # Rotar los valores del eje x
plt.show() # Imprimimos el gráfico
```
